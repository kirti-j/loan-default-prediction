```python
import pandas as pd 
import numpy as numpy

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,roc_auc_score,roc_curve,recall_score

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\Kirti\OneDrive\Desktop\Loan Default Prediction\data\processed\final_processed_data.csv")
```

```python
df.head(3)
```

```python
x=df.drop('loan_paid_back',axis=1)
y=df['loan_paid_back']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

Logistic Regression (Baseline)

```python
lr=LogisticRegression(max_iter=5000,class_weight='balanced',solver='saga')
lr.fit(x_train,y_train)
```

```python
y_pred_lr=lr.predict(x_test)
y_prob_lr=lr.predict_proba(x_test)[:,1]
```

```python
print("Logistic Regression Accuracy : ",accuracy_score(y_test,y_pred_lr)*100)
print("ROC-AUC : ",roc_auc_score(y_test,y_prob_lr)*100)
```

```python
train_acc=lr.score(x_train,y_train)*100
test_acc=lr.score(x_test,y_test)*100

print("Train score : ",train_acc)
print("Test score: ",test_acc)
```

Confusion Matrix Plot

```python
cm=confusion_matrix(y_test,y_pred_lr)
plt.figure(figsize=(5,4))
sns.heatmap(cm,annot=True,cmap='Blues',fmt='d')
plt.title("Logistic Regression -Confusion Matrix")
plt.xlabel("Prediction")
plt.ylabel("Actual")
plt.show()
```

```python
recall=recall_score(y_test,y_pred_lr)
print(f"Recall Score :{recall*100:.2f}%")
```

ROC Curve 

```python
fpr,tpr,thresholds=roc_curve(y_test,y_prob_lr)
plt.figure(figsize=(6,5))
plt.plot(fpr,tpr,label="Logistic Regression")
plt.plot([0,1,],[0,1],'--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")

plt.legend()
plt.show()
```

Random Forest

```python
rf=RandomForestClassifier(n_estimators=200,max_depth=10,class_weight='balanced',random_state=42)
rf.fit(x_train,y_train)
```

```python
y_pred_rf=rf.predict(x_test)
y_prob_rf=rf.predict_proba(x_test)[:,1]
```

```python
print("Random Forest Accuracy : ",accuracy_score(y_test,y_pred_rf)*100)
print("ROC-AUC :",roc_auc_score(y_test,y_prob_rf)*100)

print(classification_report(y_test,y_pred_rf))
```

```python
train_acc=rf.score(x_train,y_train)*100
test_acc=rf.score(x_test,y_test)*100

print("Train Accuracy : ",train_acc)
print("Test Accuracy : ",test_acc)
```

Confusion Matrix

```python
cm=confusion_matrix(y_test,y_pred_rf)
plt.figure(figsize=(6,4))
sns.heatmap(cm,annot=True,fmt='d')
plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

ROC Curve

```python
fpr,tpr,thresholds=roc_curve(y_test,y_pred_rf)
plt.figure(figsize=(6,4))
plt.plot(fpr,tpr,label="Random Forest")
plt.plot([0,1],[0,1],'--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Random Forest")
plt.legend()
plt.show()
```

XGBoost

```python
xgb=XGBClassifier(n_estimators=1000,learning_rate=0.02,max_depth=6,scale_pos_weight=4,subsample=0.8,colsample_bytree=0.8,gamma=1,reg_lambda=2,early_stopping_rounds=50,random_state=42,eval_metric='logloss')
xgb.fit(x_train,y_train,eval_set=[(x_test, y_test)],verbose=False)
```

```python
y_prob_xgb=xgb.predict_proba(x_test)[:,-1]
threshold=0.94
y_pred_xgb=(y_prob_xgb>=threshold).astype(int)
print("Applied threshold: ",threshold)
```

```python
print("XGBoost Accuracy:", accuracy_score(y_test, y_pred_xgb)*100)
print("ROC-AUC:", roc_auc_score(y_test, y_prob_xgb)*100)

print(classification_report(y_test, y_pred_xgb))
```

```python
train_acc=xgb.score(x_train,y_train)*100
test_acc=xgb.score(x_test,y_test)*100

print("Train Accuracy : ",train_acc)
print("Test Accuracy : ",test_acc)
```

```python
for t in [0.7, 0.75, 0.8, 0.85, 0.9,0.94,0.95]:
    y_pred_iter = (y_prob_xgb >= t).astype(int)
    print(f"Threshold: {t}")
    print(f"Recall for Class 0: {recall_score(y_test, y_pred_iter, pos_label=0):.2f}")
   
    print("-" * 30)
```

```python
cm=confusion_matrix(y_test,y_pred_xgb)
plt.figure(figsize=(6,4))
sns.heatmap(cm,annot=True,fmt='d')
plt.title("XGBBoost Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_prob_xgb)

plt.figure(figsize=(6,4))
plt.plot(fpr, tpr,label="XGBoost" )
plt.plot([0,1], [0,1], '--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - XGBoost")
plt.legend()
plt.show()
```

```python
y_pred_xgb=xgb.predict(x_test)
y_prob_xgb=xgb.predict_proba(x_test)[:,-1]
```

Hyperparameter Tuning

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
xgb=XGBClassifier(random_state=42,eval_metric='logloss')
```

```python
param_dist = {
    'n_estimators': [100, 200, 300,1000],
    'max_depth': [3, 5,6, 7],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.7, 0.8, 1.0],
    'colsample_bytree': [0.7, 0.8, 1.0]
}
```

```python
random_search = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_dist,
    n_iter=20,
    scoring='roc_auc',
    cv=3,
    verbose=2,
    random_state=42,
    n_jobs=-1
)
```

```python
random_search.fit(x_train, y_train)
```

```python
print("Best Params:", random_search.best_params_)

best_xgb = random_search.best_estimator_
```

```python
final_model=XGBClassifier(n_estimators=1000,learning_rate=0.02,max_depth=6,scale_pos_weight=4,subsample=0.8,colsample_bytree=0.8,gamma=1,reg_lambda=2,early_stopping_rounds=50,random_state=42,eval_metric='logloss')
```

```python

final_model.fit(x_train,y_train,eval_set=[(x_test, y_test)],verbose=False)
```

```python
y_prob_xgb=final_model.predict_proba(x_test)[:,-1]
threshold=0.94
y_pred_xgb=(y_prob_xgb>=threshold).astype(int)
print("Applied threshold: ",threshold)
```

```python
print("XGBoost Accuracy:", accuracy_score(y_test, y_pred_xgb)*100)
print("ROC-AUC:", roc_auc_score(y_test, y_prob_xgb)*100)

print(classification_report(y_test, y_pred_xgb))
```

```python
train_acc=final_model.score(x_train,y_train)*100
test_acc=final_model.score(x_test,y_test)*100

print("Train Accuracy : ",train_acc)
print("Test Accuracy : ",test_acc)
```
